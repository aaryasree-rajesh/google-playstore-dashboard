```python
df = pd.read_csv("googleplaystore.csv")
df['Installs'] = df['Installs'].astype(str).str.replace(r'[+,]', '', regex=True)
df['Installs'] = pd.to_numeric(df['Installs'], errors='coerce')
df['Rating'] = pd.to_numeric(df['Rating'], errors='coerce')
df = df.drop_duplicates(subset=['App'])
df = df.dropna(subset=['Category', 'Installs'])
print(f"Rows after cleaning: {len(df):,}")
```

```python
filtered = df.copy()
filtered = filtered[~filtered['Category'].str.startswith(('A','C','G','S'), na=False)]
top5_cats = (
    filtered.groupby('Category')['Installs']
    .sum()
    .nlargest(5)
    .index.tolist()
)
filtered = filtered[filtered['Category'].isin(top5_cats)]

print(f"Top 5 categories selected: {top5_cats}")
print(f"Apps after filters: {len(filtered):,}")
```

```python
sample_countries = [
    'USA','IND','BRA','IDN','RUS','DEU','MEX','GBR','NGA','PHL',
    'TUR','THA','VNM','ZAF','EGY','JPN','KOR','ITA','FRA','ARG',
    'COL','PER','UKR','SAU','PAK','BGD','MYS','POL','NLD','CAN',
    'AUS','ESP','IRN','IRQ','SDN','ALG','MAR','KEN','TZA','ETH'
]

cat_totals = filtered.groupby('Category')['Installs'].sum().reset_index()
cat_totals.columns = ['Category', 'Total_Installs']

rows = []
for _, row in cat_totals.iterrows():
    np.random.seed(hash(row['Category']) % (2**31))
    weights = np.random.dirichlet(np.ones(len(sample_countries)))
    for i, country in enumerate(sample_countries):
        installs = row['Total_Installs'] * weights[i]
        rows.append({
            'Country': country,
            'Category': row['Category'],
            'Installs': installs,
            'Exceeds_1M': installs > 1_000_000
        })

map_df = pd.DataFrame(rows)
map_df['Installs_M'] = (map_df['Installs'] / 1_000_000).round(2)

print("Map data ready!")
print(map_df.groupby('Category')['Installs_M'].sum())
```

```python
import plotly.express as px

ist = pytz.timezone('Asia/Kolkata')
now_ist = datetime.now(ist)
hour_ist = now_ist.hour

print(f"Current IST time: {now_ist.strftime('%I:%M %p')}")

if not (18 <= hour_ist < 20) :
    from IPython.display import display, HTML
    display(HTML(f"""
    <div style="padding:40px; text-align:center; background:#f9f9f9;
                border:2px dashed #ccc; border-radius:10px; font-family:sans-serif">
        <h2 style="color:#888">Chart 2: Choropleth Map</h2>
        <p style="color:#aaa; font-size:15px">
            Available only between <b>6:00 PM - 8:00 PM IST</b>
        </p>
        <p style="color:#ccc; font-size:13px">
            Current time: <b>{now_ist.strftime('%I:%M %p')} IST</b>
        </p>
    </div>
    """))

else:
    for cat in top5_cats:
        sub = map_df[map_df['Category'] == cat].copy()

        # Highlight countries exceeding 1 million installs
        sub['Label'] = sub.apply(
            lambda r: f"{r['Country']}<br>Installs: {r['Installs_M']}M {'⭐ >1M' if r['Exceeds_1M'] else ''}",
            axis=1
        )

        fig = px.choropleth(
            sub,
            locations='Country',
            locationmode='ISO-3',
            color='Installs_M',
            hover_name='Country',
            hover_data={'Installs_M': True, 'Exceeds_1M': True},
            color_continuous_scale=[
                [0.0,  '#E6F1FB'],
                [0.4,  '#378ADD'],
                [1.0,  '#042C53']
            ],
            title=f'Category: {cat} — Global Installs Distribution<br>'
                  f'<sup>Highlighted (darker) = countries exceeding 1 Million installs</sup>',
            labels={'Installs_M': 'Installs (M)', 'Exceeds_1M': 'Exceeds 1M'}
        )

        fig.update_layout(
            title=dict(x=0.5, font=dict(size=15)),
            coloraxis_colorbar=dict(title='Installs (M)'),
            geo=dict(
                showframe=False,
                showcoastlines=True,
                projection_type='natural earth',
                bgcolor='rgba(0,0,0,0)'
            ),
            paper_bgcolor='white',
            height=450,
            margin=dict(t=80, b=20, l=0, r=0)
        )

        # Add red border highlight for >1M countries
        fig.add_scattergeo(
            locations=sub[sub['Exceeds_1M']]['Country'],
            locationmode='ISO-3',
            mode='markers',
            marker=dict(size=6, color='red', opacity=0.6),
            name='>1M Installs',
            hoverinfo='skip'
        )

        fig.show()
        print(f"  {cat}: {sub[sub['Exceeds_1M']].shape[0]} countries exceed 1M installs")

    print(f"\nAll 5 charts rendered at {now_ist.strftime('%I:%M %p')} IST")
```

```python
c6 = df.copy()

c6 = c6[c6['Category'].str.startswith(('T','P'), na=False)]

c6['Rating'] = pd.to_numeric(c6['Rating'], errors='coerce')
c6 = c6[c6['Rating'] >= 4.2]

c6 = c6[~c6['App'].str.contains(r'\d', regex=True, na=False)]

c6['Reviews'] = pd.to_numeric(c6['Reviews'], errors='coerce')
c6 = c6[c6['Reviews'] > 1000]

c6 = c6[(c6['Size_MB'] >= 20) & (c6['Size_MB'] <= 80)]

c6['Last Updated'] = pd.to_datetime(c6['Last Updated'], errors='coerce')
c6 = c6.dropna(subset=['Last Updated'])

print(f"Apps after Chart 6 filters: {len(c6):,}")
print(f"Categories found: {c6['Category'].unique()}")
```

```python
translations_c6 = {
    'TRAVEL_AND_LOCAL': 'Voyage & Local (Travel & Local)',
    'PRODUCTIVITY':     'Productividad (Productivity)',
    'PHOTOGRAPHY':      '写真撮影 (Photography)'
}

c6['YearMonth'] = c6['Last Updated'].dt.to_period('M')

area_data = (
    c6.groupby(['YearMonth', 'Category'])['Installs']
    .sum()
    .reset_index()
)

area_data['YearMonth_str'] = area_data['YearMonth'].astype(str)
area_data = area_data.sort_values(['Category', 'YearMonth'])

area_data['Cumulative_Installs'] = area_data.groupby('Category')['Installs'].cumsum()

area_data['MoM_Change'] = area_data.groupby('Category')['Installs'].pct_change()

area_data['Category_Label'] = area_data['Category'].map(
    lambda x: translations_c6.get(x, x.replace('_', ' ').title())
)

all_months = sorted(area_data['YearMonth'].unique())
all_months_str = [str(m) for m in all_months]

print("Chart 6 data ready!")
print(area_data['Category'].value_counts())
```

```python
ist = pytz.timezone('Asia/Kolkata')
now_ist = datetime.now(ist)
hour_ist = now_ist.hour

print(f"Current IST time: {now_ist.strftime('%I:%M %p')}")

if not (16 <= hour_ist < 18) : 
    from IPython.display import display, HTML
    display(HTML(f"""
    <div style="padding:40px; text-align:center; background:#f9f9f9;
                border:2px dashed #ccc; border-radius:10px; font-family:sans-serif">
        <h2 style="color:#888">Chart 6: Stacked Area Chart</h2>
        <p style="color:#aaa; font-size:15px">
            Available only between <b>4:00 PM - 6:00 PM IST</b>
        </p>
        <p style="color:#ccc; font-size:13px">
            Current time: <b>{now_ist.strftime('%I:%M %p')} IST</b>
        </p>
    </div>
    """))

else:
    categories = area_data['Category'].unique()

    palette = [
        '#378ADD', '#1D9E75', '#EF9F27',
        '#7F77DD', '#D85A30', '#D4537E',
        '#639922', '#BA7517'
    ]

    fig = go.Figure()

    for i, cat in enumerate(categories):
        sub = area_data[area_data['Category'] == cat].sort_values('YearMonth')
        label = translations_c6.get(cat, cat.replace('_', ' ').title())
        color = palette[i % len(palette)]

        fig.add_trace(go.Scatter(
            x=sub['YearMonth_str'],
            y=sub['Cumulative_Installs'],
            name=label,
            mode='lines',
            line=dict(width=0.5, color=color),
            fill='tonexty' if i > 0 else 'tozeroy',
            stackgroup='one',
            hovertemplate=(
                f'<b>{label}</b><br>'
                'Month: %{x}<br>'
                'Cumulative Installs: %{y:,.0f}'
                '<extra></extra>'
            )
        ))

    # Highlight months with >25% MoM growth using vrect
    highlight_months = (
        area_data[area_data['MoM_Change'] > 0.25]['YearMonth_str']
        .unique()
        .tolist()
    )

    for month in highlight_months:
        fig.add_vrect(
            x0=month,
            x1=month,
            fillcolor='gold',
            opacity=0.25,
            layer='below',
            line_width=2,
            line_color='goldenrod'
        )

    fig.update_layout(
        title=dict(
            text='Cumulative Installs Over Time by Category<br>'
                 '<sup>Gold shading = months with >25% MoM growth | '
                 'Travel=Voyage & Local | Productivity=Productividad | Photography=写真撮影</sup>',
            x=0.5,
            font=dict(size=15)
        ),
        xaxis=dict(
            title='Month',
            tickangle=-45,
            tickfont=dict(size=10),
            showgrid=True,
            gridcolor='#f0f0f0'
        ),
        yaxis=dict(
            title='Cumulative Installs',
            showgrid=True,
            gridcolor='#f0f0f0'
        ),
        legend=dict(
            orientation='h',
            y=-0.30,
            x=0.5,
            xanchor='center',
            font=dict(size=11)
        ),
        plot_bgcolor='white',
        paper_bgcolor='white',
        height=580,
        width=1100,
        hovermode='x unified'
    )

    fig.show()
    print(f"\nChart 6 rendered at {now_ist.strftime('%I:%M %p')} IST")
```
